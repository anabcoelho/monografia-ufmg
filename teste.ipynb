````r
# 1. Instalar e carregar os pacotes necessários
if(!require(ggplot2)) install.packages("ggplot2")
if(!require(patchwork)) install.packages("patchwork")

library(ggplot2)
library(patchwork)

# 2. Criar um tema escuro customizado igual ao da imagem
tema_cyber <- theme(
  panel.background = element_rect(fill = "#222B35", color = NA),
  plot.background = element_rect(fill = "#222B35", color = NA),
  panel.grid.major = element_line(color = "#3A4654", size = 0.5),
  panel.grid.minor = element_line(color = "#2E3A46", size = 0.25),
  axis.text = element_text(color = "#A0AAB5", size = 9),
  axis.title = element_blank(), # Remove títulos individuais dos eixos para limpar o visual
  plot.title = element_text(color = "#D0D7DE", hjust = 0.5, size = 11, face = "plain"),
  axis.ticks = element_line(color = "#3A4654")
)

# 3. Função para gerar cada histograma padronizado
criar_histograma <- function(data, coluna, titulo) {
  ggplot(data, aes_string(x = coluna)) +
    # geom_histogram com fill=NA deixa as barras transparentes, destacando apenas a borda branca
    geom_histogram(bins = 30, fill = NA, color = "#D0D7DE", size = 0.4) +
    labs(title = titulo) +
    tema_cyber
}
````

````r
# 1. Carregar os pacotes
library(ggplot2)
library(patchwork)

# 2. NOVO: Tema Claro Customizado (Fundo branco/cinza claro e linhas escuras)
tema_claro <- theme(
  panel.background = element_rect(fill = "#F8F9FA", color = NA),
  plot.background = element_rect(fill = "#FFFFFF", color = NA),
  panel.grid.major = element_line(color = "#E9ECEF", linewidth = 0.5),
  panel.grid.minor = element_line(color = "#F1F3F5", linewidth = 0.25),
  axis.text = element_text(color = "#495057", size = 9),
  axis.title = element_blank(), 
  plot.title = element_text(color = "#212529", hjust = 0.5, size = 11, face = "bold"),
  axis.ticks = element_line(color = "#CED4DA")
)

# 3. Função usando o tema claro
criar_histograma <- function(data, coluna, titulo) {
  ggplot(data, aes(x = .data[[coluna]])) +  
    # Mudamos a cor da borda das barras para um cinza escuro azulado (#495057)
    geom_histogram(bins = 30, fill = NA, color = "#495057", linewidth = 0.4) + 
    labs(title = titulo) +
    tema_claro
}

# 4. Dados fictícios (apenas para gerar o gráfico)
set.seed(42)
dados_ficticios <- data.frame(
  Age = rbeta(1000, 2, 10) * 0.3,
  UPB_pct = rbeta(1000, 10, 2),
  FICO = rnorm(1000, 750, 40),
  Units = sample(c(0, 1), 1000, replace = TRUE, prob = c(0.95, 0.05)),
  Borrowers = sample(c(0, 1), 1000, replace = TRUE, prob = c(0.4, 0.6)),
  cltv = rnorm(1000, 75, 10)
)

# 5. Gerando os gráficos individuais claros
g1 <- criar_histograma(dados_ficticios, "Age", "Age%")
g2 <- criar_histograma(dados_ficticios, "UPB_pct", "UPB%")
g3 <- criar_histograma(dados_ficticios, "FICO", "FICO")
g4 <- criar_histograma(dados_ficticios, "Units", "Units")
g5 <- criar_histograma(dados_ficticios, "Borrowers", "Borrowers")
g6 <- criar_histograma(dados_ficticios, "cltv", "cltv")

# 6. Unindo os gráficos na grade de 2 linhas e 3 colunas
grade_graficos <- (g1 + g2 + g3) / (g4 + g5 + g6)

# 7. Criando o texto da lateral esquerda separadamente
texto_lateral <- wrap_elements(
  panel = textGrob(
    "Percentage of Occurance", 
    rot = 90, 
    gp = gpar(fontface = "plain", col = "#212529", fontsize = 12)
  )
)

# 8. Juntando o texto na esquerda com a grade de gráficos na direita
# O sinal '|' coloca o texto antes dos gráficos. O widths define o espaço do texto.
layout_final <- texto_lateral + grade_graficos + plot_layout(widths = c(1, 20))

# Se o R der erro dizendo que não conhece 'textGrob', adicione a biblioteca grid:
# library(grid)

# Mostrar o resultado
layout_final
````

````r
# 1. Carregar os pacotes
library(ggplot2)
library(patchwork)

# 2. Tema escuro corrigido (trocado 'size' por 'linewidth' nas linhas de grade)
tema_cyber <- theme(
  panel.background = element_rect(fill = "#222B35", color = NA),
  plot.background = element_rect(fill = "#222B35", color = NA),
  panel.grid.major = element_line(color = "#3A4654", linewidth = 0.5),
  panel.grid.minor = element_line(color = "#2E3A46", linewidth = 0.25),
  axis.text = element_text(color = "#A0AAB5", size = 9),
  axis.title = element_blank(), 
  plot.title = element_text(color = "#D0D7DE", hjust = 0.5, size = 11, face = "plain"),
  axis.ticks = element_line(color = "#3A4654")
)

# 3. Função atualizada sem aes_string() e usando linewidth
criar_histograma <- function(data, coluna, titulo) {
  ggplot(data, aes(x = .data[[coluna]])) +  # Jeito novo e correto de ler texto
    geom_histogram(bins = 30, fill = NA, color = "#D0D7DE", linewidth = 0.4) + # Usando linewidth
    labs(title = titulo) +
    tema_cyber
}

# 4. Dados fictícios para teste
set.seed(42)
dados_ficticios <- data.frame(
  Age = rbeta(1000, 2, 10) * 0.3,
  UPB_pct = rbeta(1000, 10, 2),
  FICO = rnorm(1000, 750, 40),
  Units = sample(c(0, 1), 1000, replace = TRUE, prob = c(0.95, 0.05)),
  Borrowers = sample(c(0, 1), 1000, replace = TRUE, prob = c(0.4, 0.6)),
  cltv = rnorm(1000, 75, 10)
)

# 5. Gerando os gráficos
g1 <- criar_histograma(dados_ficticios, "Age", "Age%")
g2 <- criar_histograma(dados_ficticios, "UPB_pct", "UPB%")
g3 <- criar_histograma(dados_ficticios, "FICO", "FICO")
g4 <- criar_histograma(dados_ficticios, "Units", "Units")
g5 <- criar_histograma(dados_ficticios, "Borrowers", "Borrowers")
g6 <- criar_histograma(dados_ficticios, "cltv", "cltv")

# 6. Juntando as partes
layout_final <- (g1 + g2 + g3) / (g4 + g5 + g6)

# 7. Título global corrigido (trocado href por hjust)
layout_final + plot_annotation(
  caption = "Percentage of Occurance",
  theme = theme(
    plot.caption = element_text(color = "#A0AAB5", size = 12, angle = 90, hjust = 0.5, vjust = 0.5),
    plot.background = element_rect(fill = "#222B35", color = NA)
  )
)
````
